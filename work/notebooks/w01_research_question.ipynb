## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python


# Research question:

Which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring based on observable search, content, and engagement signals?

# Unit of analysis:

The primary unit of analysis will be one content page. Each row represents a pseudonymized content item in the starter dataset.

# Decision:

The decision is which pages should receive human review first when the available content/SEO review capacity is limited.

# Action:

For a high-priority page, a reviewer could inspect the page and its available evidence and then decide whether to refresh it, improve its content, investigate CTR or engagement, expand it, protect it, prune it, or simply monitor it.

# Cost of a wrong recommendation

A false positive could waste limited editorial or SEO resources on a page that does not need attention. A false negative could cause a genuinely important page to be overlooked while its visibility, clicks, or engagement remain weak. Therefore, the ranking should be treated as a review aid rather than an automatic publishing or deletion decision.

The goal is to prioritize human attention, not to claim that the model can guarantee which pages will improve after an intervention.
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

```python
## 4) Careful words: what I can and can't claim

# What I can claim

I can use the available search, content, and engagement signals to identify patterns associated with declining content performance.

I can compare different ranking approaches and evaluate whether an ML-based approach can prioritize potentially important pages better than a simple baseline.

I can produce a ranked list of pages for human review and provide reason codes or supporting signals for why a page was prioritized.

I can use historical data to investigate which observable characteristics are associated with the outcome represented in the dataset.

# What I can't claim

I cannot claim that the model predicts Google's ranking algorithm or future search rankings with certainty.

I cannot claim that a page will recover simply because the model recommends refreshing it.

I cannot claim that the relationships found in the data are causal. A declining trend could have multiple explanations, including changes in search demand, competition, seasonality, SERP changes, or other factors not represented in the dataset.

I cannot claim that every page classified as declining has a content-quality problem.

I also cannot assume that the results from this 30,000-row starter dataset will automatically generalize to every website or to future data without further validation.

Therefore, the model should be treated as a **decision-support and prioritization tool**, with final actions reviewed by a human rather than automatically applied.
```
